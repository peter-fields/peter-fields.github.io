Three things that bite on the way back from Julia:
- **0-indexed.** `lst[0]` is the first element.
- **Slices are end-exclusive.** `lst[1:4]` gives 3 elements: indices 1, 2, 3.
- **`/` gives a float, `//` is integer division.** `7/2 == 3.5`, `7//2 == 3`.

## 1. Lists

```python
lst = [10, 20, 30, 40, 50]

print(lst[1], lst[-1])        # first, last
print(lst[1:4])               # end-exclusive -> [20, 30, 40]
print(lst[:2], lst[3:])       # from start / to end
print(lst[::-1])              # reversed copy
print(lst[::2])               # every other
print(len(lst), 34 in lst)
```

```python
lst = [10, 20, 30]

lst.append(70)          # add one to the end
lst.extend([80, 60])    # add many
lst.insert(0, 5)        # insert at index
print(lst)

x = lst.pop()           # remove + return last
y = lst.pop(0)          # remove + return at index
lst.remove(10)          # remove first matching VALUE (ValueError if absent)
print(lst, "popped:", x, y)

print(lst.index(20))    # index of a value
print([1, 2] + [3,4,5], [0] * 10)
```

```python
words = ["pear", "apple", "fig", "Banana"]

print(sorted(words))                        # new list
print(sorted(words, key=str.lower))         # case-insensitive
print(sorted(words, key=len, reverse=True)) # longest first

nums = [3, 1, 2]
nums.sort()             # IN PLACE -- returns None!
print(nums)

# tuple key = sort by first, break ties with second
pairs = [("c", 3), ("b", 2), ("a", 1)]
print(sorted(pairs, key=lambda p: (-p[1], p[0])) )   # count desc, name asc
```

```python
sort_pairs=sorted(pairs)
```

```python
nums = [1, 2, 3, 4, 5, 6]

print([n**2 for n in nums])                    # map
print([n for n in nums if n%2 == 0])          # filter
print([n if n % 2 else 0 for n in nums])        # conditional expression
print([(i, n) for i, n in enumerate(nums[:3])]) # nested/tuple

print(sum(nums), min(nums), max(nums))
print(any(n > 5 for n in nums), all(n > 0 for n in nums))
```

## 2. Tuples

Immutable lists. You mostly meet them as **return values** and **dict keys**.

```python
t = (1, "a", 3.0)
print(t[0], len(t))

# unpacking
x, y, z = t
print(x, y, z)

a, b = 1, 2
a, b = b, a             # swap, no temp variable
print(a, b)

first, *rest = (1, 2, 3, 4)
print(first, rest)
```

```python
# the one-element gotcha
print(type((5)) )        # int -- parens do nothing
print(type((5,)) )       # tuple -- the comma makes it

# functions return tuples
def min_max(vals):
    return min(vals), max(vals)

lo, hi = min_max([4, 9, 1])
print(lo, hi)

# immutable -> usable as a dict key
grid = {(0, 0): "origin", (1, 2): "somewhere"}
grid = {(0, 0): "origin" , (1, 2): "somewhere"}
print(grid[(1, 2)])
```

## 3. Dicts

The workhorse. Most "manage a collection of things by id" tasks are a dict.

```python
d = {"a": 1, "b": 2}
d2 = dict(a=1, b=2)                  # keyword form
d3 = dict(zip(["a", "b"], [1, 2]))   # from two sequences
# print(d == d2 == d3)

d["c"] = 3              # add or overwrite
# print(d["a"])           # KeyError if missing
# print(d.get("zz"))      # None if missing
# print(d.get("zz", 0))   # default if missing
print("d" in d, len(d))

del d["c"]
print(d.pop("b", None), d)
```

```python
d = {"apple": 3, "pear": 1, "fig": 2}

for k in d:                     # iterating a dict gives KEYS
    print(k, end=" ")
print()

for k,v in d.items() :        # the one you actually want
    print(f"{k}={v}", end=" ")
print()

print(list(d.keys()), list(d.values()))
print(sum(d.values()))
```

```python
d.items()
```

```python
d = {"apple": 3, "pear": 1, "fig": 2}

print(sorted(d.items(), key=lambda kv: kv[1]))               # by value asc
print(sorted(d.items(), key=lambda kv: kv[1])[-2:][::-1])          # top 2 by value
print({k: v for k, v in d.items() if v > 1})                 # dict comprehension
print({k.upper(): v * 10 for k, v in d.items()})
print({k.upper():v*-10 for k,v in d.items()})

# grouping: setdefault appends into a list that may not exist yet
groups = {}
for word in ["ant", "bee", "ape", "bat", "Bat"]:
    groups.setdefault(word[0], []).append(word)
print(groups)
```

#### groups

```python
from collections import Counter, defaultdict

c = Counter("mississippi")
print(c)
print(c.most_common(2))          # [('i', 4), ('s', 4)] -- already sorted desc
print(c["z"])                    # 0, not KeyError

dd = defaultdict(list)           # missing key -> empty list, automatically
for word in ["ant", "bee", "ape"]:
    dd[word[0]].append(word)
print(dict(dd))

counts = defaultdict(int)
for ch in "hello":
    counts[ch] += 1              # no need to initialise
print(dict(counts))
```

## 4. Sets

Unordered, unique, fast membership. Reach for these for dedup and "have I seen this".

```python
s = {1, 2, 3}
empty = set()           # NOT {} -- that's a dict

s.add(4)
s.discard(99)           # no error if absent (remove() would raise)
print(s, 2 in s)

print(set([1, 1, 2, 2, 3]))            # dedup
print(list(dict.fromkeys([3, 1, 3])))  # dedup, order preserved

a, b = {1, 2, 3}, {3, 4}
print(a | b, a & b, a - b)             # union, intersection, difference
```

## 6. Loops and iteration

```python
for i in range(3):              # 0, 1, 2
    print(i, end=" ")
print()
print(list(range(2, 10, 3)))    # start, stop(exclusive), step

items = ["a", "b", "c"]
for i, x in enumerate(items):           # index + value
    print(i, x, end="  ")
print()
for i, x in enumerate(items, start=1):  # 1-based, if you want it
    print(i, x, end="  ")
print()

for x, n in zip(items, [10, 20, 30]):   # walk two together, stops at shortest
    print(x, n, end="  ")
print()
```

```python
for n in [1, 3, 4, 5]:
    if n % 2 == 0:
        print("first even:", n)
        break
else:
    print("no even found")      # runs only if the loop never broke

total = 0
for n in range(10):
    if n % 3:
        continue                # skip to next iteration
    total += n
print(total)

# mutating while iterating -> use a copy, or build a new collection
d = {"a": 1, "b": 0, "c": 2}
for k in list(d):               # list() snapshots the keys
    if d[k] == 0:
        del d[k]
print(d)
```

## 7. Functions

```python
def apply_discount(price, pct=10, *, round_to=2):
    # pct is optional; round_to is keyword-only (it comes after the *)
    return round(price * (1 - pct / 100), round_to)

print(apply_discount(100))
print(apply_discount(100, 25))
print(apply_discount(100, pct=25, round_to=0))
```

```python
def summarize(*args, **kwargs):
    print("positional:", args)      # a tuple
    print("keyword:", kwargs)       # a dict

summarize(1, 2, label="x", n=3)

# THE classic bug: mutable default is created ONCE and shared
def bad(x, acc=[]):
    acc.append(x)
    return acc

print(bad(1), bad(2), bad(3))       # [1] [1,2] [1,2,3]  <-- surprise

def good(x, acc=None):
    if acc is None:
        acc = []
    acc.append(x)
    return acc

print(good(1), good(2))
```

## 8. Classes and methods

`self` is explicit and always the first parameter. That's the main thing to re-limber.

```python
class Item:
    def __init__(self, name, quantity=0, price=0.0):
        self.name = name              # instance attributes
        self.quantity = quantity
        self.price = price

    def add(self, n):                 # method: self first, always
        self.quantity += n
        return self.quantity          # remember to RETURN if callers need it

    def total_value(self):
        return self.quantity * self.price

    def __repr__(self):               # what you see when you print it
        return f"Item({self.name!r}, quantity={self.quantity})"


it = Item("widget", 5, 2.0)
print(it)
print(it.add(3), it.quantity, it.total_value())
print(it.name)
```

```python
class Counter2:
    created = 0                       # CLASS attribute -- shared by all instances

    def __init__(self, label):
        self.label = label            # instance attribute -- per object
        Counter2.created += 1

    @property
    def shout(self):                  # accessed like an attribute, no parens
        return self.label.upper()

    @staticmethod
    def describe():                   # no self -- just a namespaced function
        return "I count things"


a, b = Counter2("x"), Counter2("y")
print(Counter2.created, a.shout, Counter2.describe())
```

```python
class Base:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"I am {self.name}"


class Child(Base):
    def __init__(self, name, extra):
        super().__init__(name)        # run the parent's __init__
        self.extra = extra

    def greet(self):                  # override, reusing the parent
        return super().greet() + f" ({self.extra})"


print(Child("bot", "v2").greet())
print(isinstance(Child("b", "v"), Base))
```

```python
from dataclasses import dataclass, field

@dataclass
class Point:
    x: int
    y: int = 0
    tags: list = field(default_factory=list)   # mutable default, done safely

p = Point(1, 2)
print(p)                    # __init__, __repr__, __eq__ all written for you
print(p == Point(1, 2))
print(Point(5).tags)
```

## 9. try / except / raise

```python
def parse(s):
    try:
        return int(s)
    except ValueError:
        return None

print(parse("42"), parse("abc"))

try:
    {"a": 1}["b"]
except KeyError as e:
    print("KeyError, missing key:", e)
except (TypeError, IndexError):     # several types at once
    print("something else")
else:
    print("runs only if NO exception")
finally:
    print("always runs")
```

```python
# a base class lets callers catch the whole family at once
class InventoryError(Exception):
    pass

class ItemNotFoundError(InventoryError):
    pass

class InsufficientStockError(InventoryError):
    pass


def take(stock, sku, n):
    if sku not in stock:
        raise ItemNotFoundError(f"no such sku: {sku}")
    if n > stock[sku]:
        raise InsufficientStockError(f"want {n}, have {stock[sku]}")
    stock[sku] -= n                 # only mutate AFTER the checks pass
    return stock[sku]


stock = {"a": 5}
print(take(stock, "a", 2))

for args in [("zz", 1), ("a", 99)]:
    try:
        take(stock, *args)
    except InventoryError as e:         # catches both subclasses
        print(type(e).__name__, "->", e)

print(stock)                            # unchanged by the failures
```

```python
# common built-ins, and what triggers them
checks = [
    (lambda: [1, 2][5],        "IndexError"),
    (lambda: {"a": 1}["b"],    "KeyError"),
    (lambda: int("abc"),       "ValueError"),
    (lambda: "a" + 1,          "TypeError"),
    (lambda: None.foo,         "AttributeError"),
    (lambda: 1 / 0,            "ZeroDivisionError"),
]
for fn, expected in checks:
    try:
        fn()
    except Exception as e:
        print(f"{type(e).__name__:20} (expected {expected})")
```

```python
# 1. Return a dict mapping each word to its length.
def word_lengths(words):
    d = {}
    for w in words:
        d[w] = len(w)
    return d

assert word_lengths(["ab", "c"]) == {"ab": 2, "c": 1}
assert word_lengths([]) == {}
```

```python
# 2. Return the n most common items as (value, count), highest first, ties broken
#    alphabetically by value.
def top_n(values, n):
    vc_pairs = []
    for v in set(values):
        vc_pairs.append( (v, values.count(v)) )
    vc_pairs.sort(key= lambda p: (-p[1], p[0]) )
    return vc_pairs[:n]

assert top_n(["a", "b", "a", "c", "b", "a"], 2) == [("a", 3), ("b", 2)]
assert top_n(["b", "a"], 5) == [("a", 1), ("b", 1)]
assert top_n([], 3) == []
```

```python
# 5. A small class. Note what each method has to RETURN, and that a failed
#    withdrawal must leave the balance untouched.
class InsufficientFunds(Exception):
    pass

class Account:
    pass


acct = Account("alice", 100)
assert acct.balance == 100
assert acct.deposit(50) == 150
try:
    acct.withdraw(999)
except InsufficientFunds:
    pass
else:
    raise AssertionError("should have raised")
assert acct.balance == 150          # failed withdrawal changed nothing
assert acct.withdraw(50) == 100
assert repr(acct) == "Account('alice', balance=100)"
```

```python
# 6. Full sweep: a tiny catalogue.
#    - add(sku, name, qty): new sku -> store it; existing sku -> add to its qty.
#    - remove(sku, qty): raise KeyError if unknown, ValueError if not enough stock.
#    - low_stock(threshold): skus with qty strictly below it, sorted alphabetically.
#    - total(): total number of units held.
class Catalogue:
    pass


c = Catalogue()
c.add("b", "bolt", 5)
c.add("a", "axle", 2)
c.add("b", "bolt", 3)
assert c.total() == 10
assert c.low_stock(5) == ["a"]
c.remove("b", 8)
assert c.low_stock(5) == ["a", "b"]
assert c.total() == 2
for bad, err in [(("zz", 1), KeyError), (("a", 99), ValueError)]:
    try:
        c.remove(*bad)
    except err:
        pass
    else:
        raise AssertionError(f"expected {err.__name__}")
assert c.total() == 2
```
